## 1) make a new column for episode count

```python
data["episode_count"] = (
    data["Title"]
    .str.extract(r"\((\d+)\s+eps\)")
    .astype("Int64")
)
data.head()
```

## 2) #make a new column for time stamp


```python
data["timestamp"] = data["Title"].str.extract(
    r"([A-Za-z]{3}\s\d{4}\s-\s[A-Za-z]{3}\s\d{4})"
)
data.head()
```

#3)converting the timestaps in to the total months



```python
data["Months"] = (
    pd.to_datetime(data["Title"].str.extract(r"([A-Za-z]{3}\s\d{4})\s-\s([A-Za-z]{3}\s\d{4})")[1], format="%b %Y")
    .dt.to_period("M")
    - pd.to_datetime(data["Title"].str.extract(r"([A-Za-z]{3}\s\d{4})\s-\s([A-Za-z]{3}\s\d{4})")[0], format="%b %Y")
    .dt.to_period("M")
).apply(lambda x: x.n)
data.head()
```

## 4)give me top 5 highest scoring anime




```python
data.nlargest(5, "Score")
```

#which anime has the highest episode count


```python
data.loc[data["episode_count"].idxmax(), ["Title", "episode_count"]]
```

#animes with top 5 episode count

```python
data.nlargest(5, "episode_count")
```

#which is the Longest running anime

```python
data.loc[data["Months"].idxmax(), ["Title", "episode_count"]]
```
